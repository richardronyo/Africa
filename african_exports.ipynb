In this notebook I will be exploring African Export data provided by The Observatory of Economic Complexity (https://oec.world/en).

As a Computer Science and Mathematics grad, my expertise in Economics is limited, but I do not let that stifle my curiosity.

```python
import pandas as pd
import matplotlib.pyplot as plt

from oec import OECClient
```

```python
#I will be retrieving all of the exports from Africa and then filtering by continent
oec_client = OECClient()

continent_codes = oec_client.get_cube_members(35, 'Exporter Continent')['members']

print("Continent Codes:")
for code in continent_codes:
    print(f"\t{code['key']}: {code['caption']}")
```

```python
#Getting the Section Categories
section_categories = oec_client.get_cube_members(35, 'Section')['members']

for section in section_categories:
    print(f"\t{section['key']}: {section['caption']}")
```

```python
#Getting the HS2 Categories
hs2_categories = oec_client.get_cube_members(35, 'HS2')['members']

for i in range(len(hs2_categories)):
    print(f"\t{i + 1}: {hs2_categories[i]['caption']}")
```

```python
#Getting the HS4 Categories
hs4_categories = oec_client.get_cube_members(35, 'HS4')['members']

for i in range(len(hs4_categories)):
    print(f"\t{i + 1}: {hs4_categories[i]['caption']}")
```
